```python
def my_function(fname,lname):
    print('hello',fname,lname)

person = {'fname':'emil','lname':'refsnes'}
my_function(**person)
```

```python
def my_decorator(func):
    def wrapper():
        print("Hello")
        func()
    return wrapper

@my_decorator
def say_hi():
    print("Sahil")

say_hi()
```
