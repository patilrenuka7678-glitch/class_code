# decorators

```python
def decorator(func):
    def wrapper(*args,**kwargs):
        print('start')
        result = func(*args,**kwargs)
        print('End')
        return result
    return wrapper

@decorator
def add(a,b):
    return a + b

print(add(2,3))
```

```python
def func(**kwargs):
    print(kwargs['name'])
func(name = 'renuka',age = 27)
```

# Generator

```python
def gen():
    print('Start')
    yield 10
    print('Middle')
    yield 20
    print('End')

g = gen()
print('A')
print(next(g))
print('B')
```

```python
def gen():
    for i in range(5): # 0,1,2,3,4
        yield i

g = gen()
print(next(g)) # 0
print(next(g)) # 1
print(list(g)) # [2,3,4]
```

```python
def child():
    yield 1
    yield 2

def parent():
    yield 0
    yield from child() # child function se data ko exatract karna hai........
    yield 3

print(list(parent()))
```

```python
def gen():
    x = yield 10
    yield x * 2

g = gen()
print(next(g))
print(g.send(7))
```

```python
def gen():
    x = yield 'renuka' # this will print 1st...
    yield x * 2 

g = gen()  # genarator object creating here...
print(next(g)) # then call this...
print(g.send('patil')) # send this value to paused yield argument...
```

```python
def gen():
    x = yield '10'
    yield x * 2

g = gen()
print(next(g))
print(g.send(7))
```

```python
def gen():
    x = yield 1
    yield x

g = gen()
print(g.send(5)) # here type-error occures because of we send the value before start the generator....
```

```python
def gen():
    try:
        yield 1
        yield 2
    finally:
        print('cleanup')
        
g = gen()
print(next(g))
g.close()
print(next(g,'done'))
```
