# Logging

```python
import logging
```

```python
logging.basicConfig(filename = 'day1.log',level = logging.INFO,
                     format = "%(asctime)s - %(levelname)s - %(message)s")
```

```python
logging.info("1st project start here.")
try:
    a = int(input('Enter the 1st number: '))
    b = int(input('Enter the 2nd number: '))
    logging.info(f'we have 1st input {a} and 2nd input {b}.')
    result = a / b
    logging.info(f'we have input as {a} and  {b} which result is {result}.')
    print(result)
except Exception as e:
    logging.error(f"you'r inter a as {a} and b as {b} so this is not valid")
    print(e)
finally:
    print('Program End here.')
```

```python
logging.shutdown()
```

```python
def sqaure(a):
    logging.info('This is basic program for sqaure.')
    try:
        isinstance(a, int):
            sqaure = a * a
            logging.info(f'the sqaure of the {a} is : {sqaure} ')
    except Exception as e:
        logging.error(f'here is not valid input')
        print('Error occured:', e)
```
