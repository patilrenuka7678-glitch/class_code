Class is a blank blueprint,template or design which does not hold the any real data

# 1) Encapsulation
[ public, private, protected ]

```python
class car():
    def __init__(self,engine,fuel,seat,color):
        self.engine = engine   # this is a public so we can access it any where
        self._fuel = fuel      # This is a protected so we can access it only when we know which attribute is protected
        self.__seat = seat     # This is private so no one can access it 
        self.color = color
```

```python
ford = car(2000,'disel',4,'black')
```

```python
ford.color   # we can easily access the color and engine becuase of they are public
```

```python
ford.engine
```

```python
ford._fuel     # but when we want to access the protected attribute so we use (_) before attribute anem
```

```python
ford._car__seat      # when we want to access the private attribute so syntax = obj._classname__attribute
```

# 2) Inheritance
where are one class(parentclass) properties and attribute are access by another class (Childclass)

```python
class vehicle():
    def color(self):
        return 'this color method is belogs to the vehicle class'
```

```python
class car1(vehicle):    # here we access the parent class
    def white(self):
        return 'the color of car is white'
```

```python
toyato = car1()
```

```python
toyato.white()
```

```python
toyato.color()
```

```python
class animal():
    def sound(self):
        return 'Every animals sound different'
```

```python
class dog(animal):
    def bark(self):
        return 'Dog doing bow bow'
```

```python
dog1 = dog()
```

```python
print(dog1.sound())
print(dog1.bark())
```

# Multiple Inheritance

```python
class father():
    def driving(self):
        return 'dad can drive car smoothly'
```

```python
class mother():
    def cooking(self):
        return 'mom cook like chef'
```

```python
class child(father,mother):
    pass
```

```python
son = child()
```

```python
print(son.driving())
print(son.cooking())
```

# multilevel Inheritance

```python
class grandfather():
    def house(self):
        return 'grandfather build biggest house'
```

```python
class father(grandfather):
    def car(self):
        return 'father purchese brand new car'
```

```python
class son(father):
    def mobile(self):
        return 'son buy mobile'
```

```python
child = son()
```

```python
print(child.house())
print(child.car())
print(child.mobile())
```

# Hierarchical

```python
class animal():
    def eat(self):
        return 'what animal like?'
```

```python
class dog(animal):
    def biscuit(self):
        return 'dog like a biscuits'
```

```python
class cat(animal):
    def milk(self):
        return 'cat like a milk'
```

```python
bark = dog()
```

```python
meow = cat()
```

```python
print(bark.eat())
print(bark.biscuit())
```

```python
print(meow.eat())
print(meow.milk())
```
