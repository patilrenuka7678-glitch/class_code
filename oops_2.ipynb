# Polymorphism

```python
class dog():
    def sound(self):
        return 'dog do bow bow'
```

```python
class cat():
    def sound(self):
        return 'cat do meow meow'
```

```python
dog1 = dog()
```

```python
cat1 = cat()
```

```python
dog1.sound()
```

```python
cat1.sound()
```

```python
class car():
    def move(self):
        return 'Car move fast'
```

```python
class bike():
    def move(self):
        return 'Bike move slow than car'
```

```python
car1 = car()
```

```python
bike1 = bike()
```

```python
car1.move()
```

```python
bike1.move()
```

# Abstraction

```python
from abc import ABC, abstractmethod
```

```python
class car(ABC):
    @abstractmethod
    def airbag(self):
        pass
```

```python
class audi(car):
    def airbag(self):
        return 'airbag is available in the audi'
    def button_on(self):
        return 'car is start'
```

```python
obj = audi()
```

```python
obj.airbag()
```

```python
obj.button_on()
```

```python
class payment(ABC):
    def __init__(self,amount):
        self.amount = amount
    @abstractmethod
    def pay(self):
        pass
```

```python
class creaditcard(payment):
    def pay(self):
        return f'{self.amount} $ Pay using creadit card'
```

```python
class UPI(payment):
    def pay(self):
        return f'{self.amount} $ Pay using UPI'
```

```python
c = creaditcard(4000)
```

```python
c.amount
```

```python
c.pay()
```

```python
u = UPI(5000)
```

```python
u.amount
```

```python
u.pay()
```
