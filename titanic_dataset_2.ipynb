```python
df['Family_size'] = df['SibSp'] + df['Parch'] + 1
```

```python
df.head()
```

```python
df['is_alone'] = (df['Family_size'] == 1)    # it will return true if the passenger is alone or false if passenger with their family
```

```python
df.head()
```

```python
def age_group(age):
    if age <= 10:
        return 'Child'
    elif age <= 25:
        return 'Young'
    elif age <= 45:
        return 'Adult'
    elif age <= 75:
        return 'Senior'
    else:
        return 'Old'

df['Age_group'] = df['Age'].apply(age_group)
```

```python
df.head()
```

```python
def fare_category(fare):
    if fare <= 20:
        return 'Low'
    elif fare <= 100:
        return 'Medium'
    else:
        return 'High'
df['Fare_category'] = df['Fare'].apply(fare_category)
```

```python
df.head()
```

```python
df.groupby('Fare_category')['Survived'].mean()
```

```python
df['Cabin_know'] = df['Cabin'].notnull()  # if passenger have cabin so return true or return false if the value of cabin is NaN.
```

```python
df.head()
```

```python
df['Cabin_known'] = df['Cabin'].apply(lambda x: 'Known' if x == x else 'Unknown')
```
